# Read Data

```python
os.chdir('/content/drive/My Drive/NLP-Letters-V2/notebooks/')
df1 = pd.read_csv('../data/letters_2021.csv')
df2 = pd.read_csv('../data/sentence_sets_trimmed.csv', encoding='mac-roman')
df2 = df2[df2['full_text_tokens'] > 10]
```

```python
df1 = df1[['LETTERTEXT', 'LETTER_GENDER']]
df1 = df1.rename(columns={'LETTERTEXT':'full_text', 'LETTER_GENDER':'label'})
df1['full_text'] = df1['full_text'].str.lower()
```

```python
df2 = df2[['full_text', 'applicant_gender']]
df2 = df2.rename(columns={'applicant_gender':'label'})
df2['full_text'] = df2['full_text'].str.lower()
```

```python
df = pd.concat([df1, df2], ignore_index=True)
```

# Get Names

```python
def capitalize_names(text):
    return ' '.join(
        word.capitalize() if word.lower() in name_list else word
        for word in text.split()
    )
```

```python
df2['text_with_names_capitalized'] = df2['full_text'].apply(capitalize_names)
texts = df2['text_with_names_capitalized'].tolist()
```

```python
ner_pipeline = pipeline("ner", model="dslim/distilbert-NER", grouped_entities=True)
```

```python
results = ner_pipeline(texts, batch_size=128)
```

```python
df2['names'] = [
    [ent['word'].replace("##", "").strip() for ent in item if ent['entity_group'] == 'PER']
    for item in results
]
```

```python
names_found = df2[df2['names'].apply(lambda x: len(x) > 0)]['names']
names_found = names_found.tolist()
names_found = [name for sublist in names_found for name in sublist]
names_found = [name.lower() for name in names_found]
names_found = [name.split() for name in names_found]
names_found = [name for sublist in names_found for name in sublist]
names_found = [name for name in names_found if name in name_list]
names_found = [name for name in names_found if len(name) > 2]
names_found = list(set(names_found))
```
