```python
def replace_with_regex(text):
    for pattern, replacement in names_dictionary.items():
        text = re.sub(
            pattern,
            lambda m: replacement + (m.groupdict().get('suffix') or ''),
            text,
            flags=re.IGNORECASE
        )
    return text

df['full_text'] = df['full_text'].apply(replace_with_regex)
```
