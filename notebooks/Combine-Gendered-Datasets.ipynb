```python
os.chdir('/content/drive/My Drive/NLP-Letters-V2/notebooks/')
df1 = pd.read_csv('../data/letters_2021.csv')
df2 = pd.read_csv('../data/sentence_sets_trimmed.csv', encoding='mac-roman')
df2 = df2[df2['full_text_tokens'] > 10]
```

```python
df1 = df1.rename(columns={'LETTERTEXT':'full_text', 'LETTER_GENDER':'label'})
df1 = df1[['full_text', 'label']]
```

```python
df2 = df2.rename(columns={'applicant_gender':'label'})
df2 = df2[['full_text', 'label']]
```

```python
df = pd.concat([df1, df2], axis=0, ignore_index=True)
df['full_text'] = df['full_text'].str.lower()
```

```python
gender_label_mapping = {
    'F':0,
    'female':0,
    'M':1,
    'male':1
}

df['label'] = df['label'].replace(gender_label_mapping)
```

```python
names_dictionary = {
  "(?:^|\\b|[^\\w\\s])(?P<token>julie)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>alice)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>dylan)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>paige)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>jeanine)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>sheryl)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>jereen)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name',
  "(?:^|\\b|[^\\w\\s])(?P<token>jenn)(?P<suffix>'s|’s)?(?=\\b|[^\\w\\s]|$)": 'first_name'}
}
```

```python
def replace_with_regex(text):
    for pattern, replacement in names_dictionary.items():
        text = re.sub(
            pattern,
            lambda m: replacement + (m.groupdict().get('suffix') or ''),
            text,
            flags=re.IGNORECASE
        )
    return text

df['full_text'] = df['full_text'].apply(replace_with_regex)
```
