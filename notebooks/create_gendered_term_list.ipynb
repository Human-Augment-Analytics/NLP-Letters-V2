```python
term_dict = {}
with open("../data/term_dict.txt", "r", encoding="utf-8") as f:
    for line in f:
        key, value = line.strip().split("\t")
        term_dict[key] = value
```

```python
with open("../data/male_word_file.txt", "r", encoding="utf-8") as f_male, open("../data/female_word_file.txt", "r", encoding="utf-8") as f_female:
    male_words = [line.strip() for line in f_male]
    female_words = [line.strip() for line in f_female]
```

```python
for male, female in zip(male_words, female_words):
    if male not in term_dict:
        term_dict[male] = female
```
