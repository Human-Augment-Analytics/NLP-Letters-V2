```python
with open("../data/male_word_file.txt", "r", encoding="utf-8") as f_male, open("../data/female_word_file.txt", "r", encoding="utf-8") as f_female:
    male_words = [line.strip() for line in f_male]
    female_words = [line.strip() for line in f_female]
```
